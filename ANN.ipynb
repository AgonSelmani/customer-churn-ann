# Importing the dataset

```python
import numpy as np
import pandas as pd
import tensorflow as tf
```

# PART 1 - DATA PREPROCESSING

# Importing the dataset

```python
dataset = pd.read_csv(r"Churn_Modelling.csv")
X = dataset.iloc[:, 3:-1].values
y = dataset.iloc[:, -1].values
```

# Explaining the dataset

```python
dataset.head()

# About the dataset:
    # This is a dataset from a bank which collected some informations about their customers as we can see
    # Each row is a Customer
    # 'Exited' is the dependet variable where: 1 means the customer exited this bank, 0 means he/she is still a customer.

    # This bank observed their customers for a certain period of time (like 6 months or more) and got this results during this time.
    # And want to understand the correlation between those features and the fact that whether customer stays in this bank or leaves the bank.

    # That makes sense because the bank wants to have customers as much as possible.
    # So they want to keep the maximum customers thats why they want to understand WHY THE CUSTOMERS LEAVE THE BANK ?
    # Once they manage to bulid a predictive model that can predict if any new customer leaves the bank,
    # and for those costumers the bank will be prepared and might do some special offers to them so they stay in the bank.


# So our job is to train a predictive model on this dataset to understand the correlations between all of these features and the dependet variable
# and to deploy our model in future customers to do the prediction if this customer will stay in the bank or not.
# AND
# Even better than this we will predict the probability that this customer leaves the bank.
```

# Encoding categorical data

```python
# We know before hand that the dataset does not have missing data so we can move to encoding

# Label encoding the "Gender" column
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X[:, 2] = le.fit_transform(X[:, 2])

# One hot encoding the "Geography" column
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [1])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

```python
print(X)
```

# Spliting into training set and test set

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
```

# Feature Scaling

```python
# Feature Scaling is really important for ANN
# So important that we need to do it on all the columns

from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

# PART 2 - BUILDING THE ANN

# Intializing the ANN

```python
ann = tf.keras.models.Sequential() #Intialize the Neural Network as a sequence of layers
```

# Adding the input layer and the first hidden layer

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu')) 
# 6 neuron in the first hidden layer, we get this number by experimenting, 
# in the input layer we would have as much input neurons as there are features 
# rectifier activation function 
```

# Adding the second hidden layer

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu')) 
# The same, if ofcurse we want the same number of neurons
```

# Adding the output layer

```python
ann.add(tf.keras.layers.Dense(units=1, activation='sigmoid')) 
# Because we have binnary classification we need just one neuron
# If we would have answers like A/B/C we would have 3 neurons for like one hot encoding (ex: A = 1:0:0, B = 0:1:0, C = 0:0:1).

# Sigmoid because we also get the probabilites that the outcome is 1.
```

# PART 3 - TRAINING THE ANN

# Compiling the ANN

```python
ann.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
# Does Stochastic gradient Descent for updating the weights
# For nonbinary loss = 'classification_crossentropy' and activation function should not be sigmoid but softmax
# loss is the way (function, formula) to calculate the difference between the prediction and the real results
```

# Training the ANN on the Training set

```python
ann.fit(X_train, y_train, batch_size = 32, epochs = 100)
# batch_size gives you the number of prediction you want to compare to the same number of real results at a time
# so we are doing batch learning.
```

# PART 4 - MAKING THE PREDICTIONS AND EVALUATING THE MODEL

# Predicting the result of a single observation

```python
print(ann.predict(sc.transform([[1,0,0,600,1,40,3,60000,2,1,1,50000]])) > 0.5)
```

# Predicting the test set result

```python
y_pred = ann.predict(X_test)
y_pred = (y_pred > 0.5)

print(np.concatenate((y_pred.reshape(len(y_pred), 1), y_test.reshape(len(y_test), 1)), 1)) 
# with reshape we are reshaping the vectors to "vector: len(vector) X 1", which means instead of horizontally we want the vectors vertically 
# second argument of concatenate: 1, means we want to concatenate vertically, if it was 0 it would mean horizontally.
```

# Making the confusion matrix

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
cm = confusion_matrix(y_test, y_pred)
ac = accuracy_score(y_test, y_pred)
print(cm)
print("The accuracy: ",ac)
```
